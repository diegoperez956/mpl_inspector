Run the next cell, then hover and click inside the figure.

If the backend printed above is not `module://ipympl.backend_nbagg` or some `nbAgg` variant, the plot may render but hover/click inspection will not be interactive.

- hover to preview

- `t` to print the artist tree


````python
rng = np.random.default_rng(7)
x = np.linspace(0, 2 * np.pi, 250)

with plt.ioff():
    fig, axes = plt.subplots(2, 2, figsize=(11, 7), constrained_layout=True)

fig.suptitle("mpl_inspector notebook demo")
line_ax, scatter_ax, bar_ax, image_ax = axes.flat

line_ax.plot(x, np.sin(x), label="sin(x)", linewidth=2.5)
line_ax.plot(x, np.cos(x), label="cos(x)", linewidth=2.5, linestyle="--")
line_ax.set_title("Lines")
line_ax.legend(loc="upper right")
line_ax.annotate("peak", xy=(np.pi / 2, 1.0), xytext=(2.0, 1.3), arrowprops={"arrowstyle": "->"})

sx = rng.normal(size=70)
sy = 0.6 * sx + rng.normal(scale=0.55, size=70)
scatter_ax.scatter(sx, sy, c=np.hypot(sx, sy), s=rng.uniform(40, 180, size=70), cmap="viridis", alpha=0.8, label="samples")
scatter_ax.axhline(0, color="#94a3b8", linewidth=1.0)
scatter_ax.axvline(0, color="#94a3b8", linewidth=1.0)
scatter_ax.set_title("Scatter")
scatter_ax.legend(loc="upper left")

bars = bar_ax.bar(["alpha", "beta", "gamma", "delta"], [3.2, 5.7, 2.4, 6.4], color=["#60a5fa", "#f59e0b", "#34d399", "#f472b6"], label="bars")
bars[1].set_hatch("//")
bar_ax.axhspan(4.5, 5.5, color="#fde68a", alpha=0.25)
bar_ax.set_title("Bars")
bar_ax.legend(loc="upper left")

matrix = rng.normal(size=(18, 18))
img = image_ax.imshow(matrix, cmap="magma", interpolation="nearest")
image_ax.set_title("Image")
image_ax.text(2, 15, "Inspect me", color="white", fontsize=11)
fig.colorbar(img, ax=image_ax, shrink=0.78)

inspector = mpl_inspector.display(fig, print_on_select=False)
````

````python
inspector.dump_tree()
````
